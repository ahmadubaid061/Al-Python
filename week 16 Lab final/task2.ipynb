```python
import pandas as pd
df=pd.read_csv('./Mall_Customers - Mall_Customers.csv')
df.head()
```

Feature Selection

I think only four ('Gender','Age','Annual Income (k$)','Spending Score (1-100)')  are suitable features for kmeans cause these can be efficient in classifying data to many clusters
ID column has notjing to do with classification

```python
X=df[['Gender','Age','Annual Income (k$)','Spending Score (1-100)']]
```

Encoding Gendere
because it is string column and cannot be used directly in kmeans

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
X['Gender']=le.fit_transform(X['Gender'])
```

Feature Scalling

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X['Annual Income (k$)']=scaler.fit_transform(X[['Annual Income (k$)']])
X['Spending Score (1-100)']=scaler.fit_transform(X[['Spending Score (1-100)']])
X.head()
```

Checking for missing values

```python
X.isna().sum()
```

training model

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init=10)
X['Clusters'] = kmeans.fit_predict(X)
```
