## multyple plots in a single figure


```python
import matplotlib.pyplot as plt
import numpy as np

figure,ax=plt.subplots(2,2,figsize=(8,6))
# the arguments 2,2 means that there should be two rows and two columns in the figure
# means total 4 plots can be plotted here

### plot 1 line graph
x=np.linspace(0,10,100)  #0 = mean,  10=standard deviation, 100 total samples
y=np.sin(x)             # sin of each value of x

ax[0,0].plot(x,y,linewidth=1,alpha=0.8,color='b',marker='o',markersize=1.5)
ax[0,0].set_title("Line Plot")


## plot 2 scatter graph
ax[0,1].scatter(x,y,alpha=0.8,color="b",marker='*')
ax[0,1].set_title("Scatter Plot")

## plot 3 Histogram
ax[1,0].hist(np.random.randn(200),bins=20,color='g')
ax[1,0].set_title("Histogram")

## Plot 4 bar Plot
categories=["A","B","C","D"]
values=[1,2,3,4]
ax[1,1].bar(categories,values,color="m")
ax[1,1].set_title("Bar Plot")


plt.tight_layout()
plt.show()
```
