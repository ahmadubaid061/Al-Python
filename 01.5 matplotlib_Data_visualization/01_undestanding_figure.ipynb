## Matplotlib Figure Understanding

```python
import matplotlib.pyplot as plt
figure,ax=plt.subplots()  #this is an object which return the figure and an array of axis

# labeling the axis
ax.set_title("Blank figure with matplotlib")
ax.set_xlabel("this is the x axis label")
ax.set_ylabel("this is the y axis label")

# saving the figure
figure.savefig("blank_figure.png",dpi=100)  #dpi means pixels density

#if using just .py file then you need figure.show() function to see the output
figure.show()
```
