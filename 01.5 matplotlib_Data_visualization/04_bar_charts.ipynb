## Bar Charts
- used for grouped data

```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv("ecommerce_dataset.csv")
df.head()
```

## grouping data

```python
# grouping the products by category and then plotting the revenue of each caategory
category_revenue=df.groupby("product_category")['revenue'].sum()
category_revenue
```

## plotting the figure

```python
figure,ax=plt.subplots()
ax.bar(category_revenue.index,category_revenue.values,label="category revenue")

ax.set_title("product category revenue")
ax.set_xlabel("product category")
ax.set_ylabel("category revenue")
ax.legend()

figure.show()
```

## Grouped Bar chart
- for category and region combined

```python
grouped_revenue=df.groupby(['product_category','region'])['revenue'].sum()
# the grouped category is a dictionary with tuples (category, region ) as keys and revenue as values
for (category, region), total_revenue in grouped_revenue.items():
    print(f"{category}->{region} : {total_revenue}")
```

### plotting the grouped revenue of cateory and region

```python
grouped_revenue.plot(kind='bar',figsize=(10,4),)
plt.title("Grouped Revenue of category and region combined")
plt.ylabel("revenue")
plt.xlabel("category + region")
plt.show()
```

## UnStacked revenue for the Grouped revenue

```python
grouped_revenue=grouped_revenue.unstack()
grouped_revenue.plot(kind='bar',figsize=(8,4))
plt.title("Grouped Revenue of category and region combined")
plt.ylabel("revenue")
plt.xlabel("category + region")
plt.show()
```

## Stacked Revenue for grouped data

```python
grouped_revenue.plot(kind='bar',stacked=True,figsize=(10,4))
plt.title("Grouped Revenue of category and region combined")
plt.ylabel("revenue")
plt.show()
```
