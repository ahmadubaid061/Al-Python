## Line Plot
- used for simple and linear data
- represents relation between two linearly dependent variables

```python
import matplotlib.pyplot as plt

# first of all we need data 
Sales=[100,150,200,250,120,300,90,200]
Days=[1,2,3,4,5,6,7,8]

# now plotting the data using a line plot
# plt.plot is used for line plot
plt.plot(Days,Sales,marker='o',linewidth=2,alpha=1,label="Daily Sales")
# explanation 
# the first two arguments are the data variables we want to plot (x-axis,y-axis)
# marker=o means that each data point should be represented as 'o'
# aplha=0.8 means the opacity or desity of the the line color
# label is the lable for plot means it has the figure info

plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()  #legend is sown in a corner of the plot which shows guidlines
plt.show()

```

## Scatter plot

```python
plt.scatter(Days,Sales,marker="*",alpha=1,label='daily sales')
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()  
plt.show()
```
