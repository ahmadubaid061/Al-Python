## Histogram
- used to plot the frequency of data

```python
import matplotlib.pyplot as plt
import numpy as np
# we will use numpy to generate random samples

group_a=np.random.normal(70,10,1000)  #70= mean , 10= standard deviation , 1000 = samples
group_b=np.random.normal(80,12,1000)
print("group_a : ",group_a[:10]) #means print first 10 sampels
print("group_b : ",group_b[:10])
```

## Plotting histogram for group a and b

```python
plt.hist(group_a,bins=30,alpha=1,density=True,label="Group A")
plt.hist(group_b,bins=30,alpha=0.5,density=True,label="Group B")
# What are "bins" in a histogram?
# Bins are the intervals or "buckets" that divide your data into groups when creating a histogram. Each bin represents a range of values, and the height of each bar shows how many data points fall into that range.

plt.xlabel("score")
plt.ylabel("Density")
plt.title("Score Density Histogram")
plt.legend()
plt.show()
```

## Histogram for ecommerce dataset revenue

```python
import pandas as pd
df=pd.read_csv('ecommerce_dataset.csv',parse_dates=['order_date'])

figure,ax=plt.subplots(figsize=(12,4))
ax.hist(df['revenue'],bins=30,alpha=0.8)
ax.set_xlabel("revenue")
ax.set_ylabel("Frequency")
ax.set_title("Revenue Distribution")

figure.show()
```

## Density for ecommerce dataset revenue

```python
figure,ax=plt.subplots(figsize=(12,4))
ax.hist(df['revenue'],bins=30,density=True,alpha=0.8)
ax.set_xlabel("revenue")
ax.set_ylabel("Density")
ax.set_title("Revenue Distribution")

figure.show()
```

## Comparing revenue vs Discounted revenue
- we will plot histogram for revenue  discounted revenue in the same figure 
- to see the differnce 

```python
discounted_revenue=df['revenue']*(1-df['discount']) #revenue afer discount
figure,ax=plt.subplots(figsize=(10,4))
ax.hist(df['revenue'],density=True, bins=30,alpha=0.4,label='Original Revenue')
ax.hist(discounted_revenue,density=True, bins=30,alpha=1,label="Discounted Revenue",color="#333")

ax.set_title("Revenue Comparison")
ax.legend()
figure.show()
```
