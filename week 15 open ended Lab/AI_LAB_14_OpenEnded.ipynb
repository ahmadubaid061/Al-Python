**Part 1: Data Understanding and Preprocessing**

```python
import pandas as pd
df=pd.read_csv('./creditcard.csv')
print(df.shape)
df.head()
```

```python
print(df['Class'].value_counts())
df['Class'].value_counts(normalize='True')*100  #in proportion
```

```python
#classes are imbalance so need balancing
#but first we will split data#
X=df.drop(["Class"],axis=1)
Y=df['Class']
```

splitting into train_test

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)
print(X_train.shape)
print(X_test.shape)
```

**Balancing the training data**

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE(random_state=42)
X_train_bal,Y_train_bal=sm.fit_resample(X_train,Y_train)
Y_train_bal.value_counts(normalize="True")*100
```

now data is 100% balanced

**First we should scale the data**

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
X_train_bal=scalar.fit_transform(X_train_bal)
test_index = X_test.index  # save index before scaling (needed for Q4)
X_test_scaled=scalar.transform(X_test)
#now X_train_bal is a numpy array so we cannot use .head()
X_train_bal[:2] #print first two rows only
#all the values will be now in range of 0 and 1 almost
```

**Now i will be using the decision tree model to apply hard and soft constraint**

First i will find fraud probability

```python
fraud_probs = tree_model.predict_proba(X_test_scaled)[:, 1]
fraud_probs
```

Adding more columns to df for applying contraints

```python
import numpy as np
df['account_age_category'] = pd.cut(df['Time'],
    bins=[0, 50000, 100000, np.inf],
    labels=['New', 'Medium', 'Old'])     #this will create a new column account_age_category
                                         #3 categories New,Medium,Old
df.head()
```

country_mismatch

```python
np.random.seed(42)
df['country_mismatch'] = np.random.choice([0, 1], size=len(df), p=[0.8, 0.2])
#create a new column with random data 80% 0's and 20% 1's
```

Regulatory flag

```python
df['regulatory_flag'] = (df['Amount'] > 1000).astype(int)
df.head()
```

Applying Hard Constraints

```python
def apply_hard_constraints(row):
    if row['Amount'] > 1000 and row['country_mismatch'] == 1:
        return 'Escalate'
    if row['account_age_category'] == 'New' and row['Amount'] > 500:
        return 'Block'
    if row['regulatory_flag'] == 1:
        return 'Escalate'
    return None
```

Soft Contstraints

```python
def apply_soft_constraints(fraud_probability):
    if fraud_probability < 0.3:
        return 'Approve'
    elif fraud_probability > 0.7:
        return 'Block'
    else:
        return 'Review'
```

Final Decision Funciton

```python
def final_decision(row, fraud_probability):
    hard = apply_hard_constraints(row)
    if hard:
        return hard
    return apply_soft_constraints(fraud_probability)
```

creating a copy of X_test data

```python
# Convert scaled numpy array back to DataFrame with original index
X_test_copy = pd.DataFrame(X_test_scaled, index=test_index, columns=X.columns)

# Add fraud probability from decision tree model
X_test_copy["fraud_probability"] = fraud_probs

# Add the 3 engineered columns using saved test_index
X_test_copy["country_mismatch"] = df.loc[test_index, "country_mismatch"].values
X_test_copy["account_age_category"] = df.loc[test_index, "account_age_category"].values
X_test_copy["regulatory_flag"] = df.loc[test_index, "regulatory_flag"].values

X_test_copy.head()  # preview the result
```

Applying final decision function to every transaction

```python
# Apply final decision to every row in test data
X_test_copy["final_decision"] = X_test_copy.apply(
    lambda row: final_decision(row, row["fraud_probability"]), axis=1
)

# Show count of each decision
print(X_test_copy["final_decision"].value_counts())
```
