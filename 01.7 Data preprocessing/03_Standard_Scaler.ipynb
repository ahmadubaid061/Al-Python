# StandardScaler
- it is one of free defined functions for machine learning
- it uses the concept of zscore behind the scene

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df=pd.read_csv("telecom_customer.csv")
df['monthly_income'].head()
```

```python
#scalling monthly income using standard scaler
scaler=StandardScaler()
df['monthly_income_scaled']=scaler.fit_transform(df[['monthly_income']])
```

```python
# observing side by side columns
df[['monthly_income','monthly_income_scaled']].head()
```

```python
# this is it no need of calculating z-score and then using manual formula 
# just call the StandardScaler constructure and it would do it
```
