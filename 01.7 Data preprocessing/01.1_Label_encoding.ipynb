# label encoding
- label coding is the easiest encoding method
- it doesnot create a seperate column for each unique category but
- it just creates a single column where each category has its own numerical value

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('insurance_dataset.csv')
df.head()
```

```python
# for a column where the the categories are not in order we cannot use ordinal encoding
# and for a column where we have multiple values like suppose bmi cateory we should not use one hot encoding 
#because one hot encoding will create seperate columns for categories like 'Normal , Obese, Overweight, Underweight'
# we can use label encoding for thaat
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
df['bmi_category_encoded']=encoder.fit_transform(df['bmi_category'])
df.head()
```

```python
df=df.drop(['bmi_category'],axis=1)
df.head()
```
