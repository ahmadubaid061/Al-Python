# Test train Slit
- when we train a model we need to test it then
- for that perpose we do not provide our whole dataset to the model during trainining
- first we divide data into training and testing portions (80% training and 20% testing)

```python
import pandas as pd
df=pd.read_csv('data.csv')
df.head()
```

## Labeling data
- we divide columns into two types of variables 
- the x variable means independent data 
- y variable means the target column or dependant data

```python
X=df['square_footage']
Y=df['monthly_rent']
```

## Spliting data into testing and training data

```python
from sklearn.model_selection import train_test_split
# applying train_test_split
X_train, X_test, y_train, y_test=train_test_split(X,Y,test_size=0.2,random_state=42,)
# it will give us four lists of data X_train, X_test, y_train, y_test
# testsize means that 0.2(20%) data should be assigned to testing and remaining to trainingg
# randomstate shupples data so that the during splitting random data is assigned to test and train not in sequence
```

## checking the lenght of each list and comparing with oriinal columns

```python
print(f'X original length: {len(X)}')
print(f'Y original length: {len(Y)}')
print(f'X_train original length: {len(X_train)}')
print(f'X_test original length: {len(X_test)}')
print(f'y_train original length: {len(y_train)}')
print(f'y_test original length: {len(y_test)}')
```

## printin side by side to see the data 

```python
#first i will conert them to numpy arrays
import numpy as np
X_train = np.array(X_train)
X_test = np.array(X_test)
y_train = np.array(y_train)
y_test = np.array(y_test)

# Print side by side
print("X_train\ty_train\tX_test\ty_test")
print("-" * 60)

for i in range(min(10, len(X_train), len(X_test))):
    print(f"{X_train[i]}\t {y_train[i]}\t {X_test[i]}\t {y_test[i]}")
```
