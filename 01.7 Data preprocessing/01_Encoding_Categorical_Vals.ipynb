## Encoding Categorical Values
- machine learning models only work with numerical values
- those column with categorical(string dtype) should be converted to numerical data type
- for that we use Encoding
- ### Encodin is of two types
- one hot encoding
- ordinal encoding

# 1:Ordinal encoding
#### What it does: 
- Assigns a numeric rank (e.g., 0, 1, 2, 3) to categories based on their natural order or importance.

#### When to use: 
- When the categories have a clear, logical hierarchy (e.g., "Small" < "Medium" < "Large").

#### importance:
-  The numbers imply ordering but not equal spacing (e.g., the difference between 0→1 is not necessarily the same as 1→2).

```python
import pandas as pd
df=pd.read_csv('insurance_dataset.csv')
df.head()
```

```python
#in the dataset the categorical columns are bmi_category,sex,smoker,region
#sex,smoker,region these columns needs one hot encoding because there is no hierarchy
#the bmi_cateory however needs ordinal encoding because the categories are in hierarchecal form
#one category is greater than the other and ...
# so here we will encode the bmi category with ordinal encoding
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder=OrdinalEncoder(categories=[['Normal','Overweight','Obese','Underweight']])
df['bmi_categ_encoded']=encoder.fit_transform(df[['bmi_category']])
df.head() #we can see a new column 'bmi_categ_encoded' is added
```

```python
# as we do not need the original bmi_category column we should drop it now
df=df.drop(['bmi_category'],axis=1)
df.head() #the bmi_category column is now removed
```

# 2. One-Hot Encoding
#### What it does:
-  Creates a new binary column (0 or 1) for each unique category. For each row, only one of these columns gets a 1 (the category that applies), and the rest get 0.

#### When to use:
-  When the categories have no natural order (nominal data).

#### Important: 
- Avoids forcing a false ranking, but increases the number of columns.

```python
# for encoding 'sex','region','smoker' we will use one hot encoding
df=pd.get_dummies(df,columns=['sex','region','smoker'],drop_first=False) #drop_first removes the original column
# this will create two binary encoded columns for sex as 'sex_male and sex_female'
# four binary encoded columns for region as 'region_northeast,region_northwest,region_southeast,region_southwest
#two binary encoded columns for smoker as smoker_yes and smoker_no
df.head()
```

```python

# also the smoker_yes column can tell us wether the person is smoker or not so we do not need the smoker_no column
# also the sex_male column can tell us wether male or female so do not need the sex_female column
df=df.drop(['sex_female','smoker_no'],axis=1)
df.head()
```
