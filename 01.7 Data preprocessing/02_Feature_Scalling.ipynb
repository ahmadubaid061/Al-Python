## Feature Scalling 
- some times the data from different columns varies alot
- in such dataset machine learning model will pick the column with larger values and ignore the other
- to avoid such we use feature scaling
- there are two types of scalling 

## 1:Z-score scalling
- scales data so that the entire column's mean is 0

## 2:min_max scallling
- scales data between 0 and 1

```python
import pandas as pd
df=pd.read_csv('telecom_customer.csv')
# we will just observe the monthly income and bill
df[['monthly_income','monthly_bill']].head()
```

```python
# data information
df[['monthly_income','monthly_bill']].describe()
```

```python
# as we can see the values varies by 1 figure one is in 1000s and the other in 10000s
# we wll use z-score for monthly income and min_max for monthly bill
```

## Monthly income scalling with z-score
- z-score = (value-meanOFColumn)/standard_deviation

```python
income_mean=df['monthly_income'].mean()
income_std=df['monthly_income'].std()
print(income_mean,income_std)
```

```python
# applying z-score
df['monthly_zscore_income']=(df['monthly_income']-income_mean)/income_std
df[['monthly_income','monthly_zscore_income']].head()
```

```python
df=df.drop(['monthly_income'],axis=1) #no need of the original income column now
df.head(2)
```

## Scalling Monthly Bill with min_max scalling
- min_max= (value-MinValue)/(maxValue-MinValue)

```python
bill_min=df['monthly_bill'].min()
bill_max=df['monthly_bill'].max()
print(bill_min,bill_max)
```

```python
# Applying the min_max formula
df['monthly_minmax_bill']=(df['monthly_bill']-bill_min)/(bill_max-bill_min)
df[['monthly_bill','monthly_minmax_bill']].head()
```

```python
# no need of original monthly_bill column
df=df.drop(['monthly_bill'],axis=1)
df.head(2)
```

## comparing the scalled monthly income and bill

```python
df[['monthly_zscore_income','monthly_minmax_bill']].head(15)
# now the values of both columns are scalled near 0 and 1 and the model will not drop either one
```

```python
df[['monthly_zscore_income','monthly_minmax_bill']].describe()
```

#### All in the range of 0 and 1 now
