# Always keep these steps in mind for training a model

## 1: Problem Definition
- what are predicting ? (Regression,Classification,clustering)

## 2: Data Collection
- Read data from a csv or any other source

## 3: Data Pre Processing
- handle missing values (drop or fill with mean/median)
- handle outliers (remove or cap)
- Encode Categorical values
- Feature Scalling (StandardScaler)
- Features Engineering (create new features if need and drop useless)

## 4: Data splitting
- Divide data into training and testing data
- normal ratio for test train is 20% testing and 80% training data

## 5: Model Training
- Feed Data to the model ( choosen in step 1)
- Note : the model should never see the testing data before training

## 6: Model Evaluation and testing
- now use the testing data to predict target value by model 
- if classification problem use metrices like Accuracy,Recall,Precision,F1 score etc
- if Regression problem use metrices like MAE,MSE,RMSE,R2-score

## 7: Finalalize or Re-train
- if model give result that was expected it is okay 
- else if model was not giving the expected result go back and choose a different technique
- either use a different feature scalling technique or a different model
