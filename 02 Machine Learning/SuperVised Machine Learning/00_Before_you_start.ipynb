## what is supervised learning
- super vised learning is the one in which we have the target column in the dataset
- the target column is also called label 

## there are two sub types of supervised learning
- 1: Regression
- 2: Classification

### 1: Regression
- when the target column is of numerical data type the problem is regression problem
- the model that is used for sovling regression problem is;
- 1: Linear Regression
- 2: Decision Tree Regressor

### 2: Classification 
- when the target column is of categorical values (strings) the problem is called classification problem
- the models used for solving classification problem are;
- 1: Logistic Regression (binary classification)
- 2: Support Vector machine . svm (Binary Classification)
- 3: k-nearest neighbors knn (multiple classification)
- 4: Decision Trees
- 5: naive bayer's theorm
- 6: Neural Networks (Deep Learning model)
