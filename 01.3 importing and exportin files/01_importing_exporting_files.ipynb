## reading CSV files


```python
import pandas as pd
import numpy as np
students_df=pd.read_csv('students_data.csv')
print(students_df.info())   #prints all information about dataset
```

```python
print(students_df.describe())  #prints statisticall details
```

```python
print(students_df.head(10)) #only 10 rows
```

```python
# as we can see there are some null values 
print(students_df.isna().sum() )#tells us which columns contains null vlues and how much
```

```python
# now suppose i wnat to remove the null values entire rows
# dropna() function can only remove NAN values so
# we will first convert empty values to nan
students_df.replace('',np.nan)
students_df = students_df.dropna(axis=0) #axis=0 means entire row
print(students_df.head(5)) #starting five rows
print('')
print(students_df.tail(5)) #last five rows
```

## Exporting as excel file

```python
# for reading and writing excel files we need another library
# - pip install openpyxl (in case if using .py file instead of ipynb)
import openpyxl
students_df.to_excel('exported_students.xlsx',sheet_name='students_data')
print("data expored into an excel file")
```

## Exporting as json file

```python
students_df.to_json('exported_students.json',orient='records',indent=2)
# indent means margins or space between each record
print("data exported to exported_students.json ")
```
