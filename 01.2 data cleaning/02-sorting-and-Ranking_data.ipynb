## Reading data

```python
import pandas as pd
df=pd.read_csv('retail_sales.csv')
#the csv file has 1800 lines and i just need max 50 so i will take a little portion of that
df=df.head(50)
print(df.head(5))
```

#### droping null value rows

```python
df=df.dropna(axis=0)
# that will just check for numerical null values 
# if there are strings like "Nan"
# for that we should do this!
df['Category']=df['Category'].replace("Nan",'Undefined') #replaces Nan with undefined
df['Region']=df["Region"].replace("Nan",'Undefined')
print("Column   null values\n",df.isna().sum())
df.head(20)
```

## sorting by quantity

```python
sorted=df.sort_values(by='Quantity',ascending=True)
print(sorted.head(5))
```

### sorting by two combine properties

```python
sorted_df=df.sort_values(by=['Quantity','Sales'],ascending=[False,True]) 
#quantity descending and Sales ascending
print(sorted_df.head(10))
```

## Ranking Data

```python
#we will add a new column to the existing dataset named sale_rank
df['sale_rank']=df.groupby('Quantity')['Sales'].rank(method='dense',ascending=False)
print(df.head(20))
```
