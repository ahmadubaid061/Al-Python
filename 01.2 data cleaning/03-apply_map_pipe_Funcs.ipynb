## map Function

```python
# suppose i want to create new column payment type which
# will store the method of of parment
# suppose i want the card and wallet to be both considered as digitall 
df['payment_type']=df['payment_method'].map(
    {
        "Card":"Digital",
        "Wallet":"Digital",
        "COD":"Cash"
    }
)

# cross checking with patment_method
df[['payment_method',"payment_type"]].head()
```

```python
# suppose i want to calculate how much payments were digital and how much cash
df.groupby('payment_type').size()
```

## Apply Function
- Apply function is used when more than one columns are invlved
#### used:
- To apply a function that doesn't exist in Pandas
- To transform data based on complex conditions
- To create new columns from existing ones
- syntax: df.apply(function, axis=0) or axis=1

```python
# suppose i want to find total price of each product by multiplyin unit price with quantity
def total_price(row):
    return row["unit_price"]*row["quantity"]

# now i will apply the function to dataset
df['Total Amount']=df.apply(total_price,axis=1)
df.head(10) #now we have an extra column named Total Amount
```

## pipe Function
The pipe() function allows you to chain multiple functions together in a clean, readable way. It's used for method chaining - applying a series of transformations to your data in sequence.
- syntax : df.pipe(function1).pipe(function2).pipe(function3)

```python
def add_total_price(df):
    """Function 1: Calculate total price for each order"""
    df['total_price'] = df['unit_price'] * df['quantity']
    return df

def add_net_price(df):
    """Function 2: Calculate net price after discount"""
    df['net_price'] = df['total_price'] * (1 - df['discount'])
    return df

def categorize_orders(df):
    """Function 3: Categorize orders based on net price"""
    df['order_category'] = df['net_price'].apply(
        lambda x: 'High' if x > 1000 else ('Medium' if x > 500 else 'Low')
    )
    return df

# Apply pipe with three functions
df= (df.pipe(add_total_price).pipe(add_net_price).pipe(categorize_orders))
df.head() #now we have total price, net price , order category
```
