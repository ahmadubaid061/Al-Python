## creating a dataframe 
- some values contians spaces before or after
- some contains '-' as combiner should be replaced by '_'
- some are capital should be converted to lower case

```python
data={
    'Name': ['Product A', 'Product B', 'Product C', 'Product D', 'Product E', 
             'Product F', 'Product G', 'Product H', 'Product I', 'Product J',
             'Product K', 'Product L', 'Product M', 'Product N', 'Product O',
             'Product P'],
    'Category': ['Electronics', 'Home-Kitchen', 'electronics', 'CLOTHING', 
                 'Electronics', 'hand-Bag', ' BEAUTY ', 'Home-Kitchen ', 'Electronics',
                 'CLOTHING ', 'beauty', ' Electronics',  'Beauty',
                 'electronics ', ' Beauty',  'ELECTRONICS'],
}
# the category column contains those error mentioned above and we will clean those errors
```

```python
import pandas as pd
df=pd.DataFrame(data)
print(df)
```

## removing spaces before or after

```python
df['Category']=df['Category'].str.strip() #strip removes spaces before and after
df
```

## Replacing '-' with '_'

```python
df['Category']=df['Category'].str.replace('-','_')
print(df) # now the - every where is replaced with _
```

## Converting to lower case

```python
df["Category"]=df['Category'].str.lower()
print(df)
```
