```python
import pandas as pd
df=pd.read_csv('ecommerce_dataset.csv')
df.head()
```

- #### First i need to find the total net price for each order

```python
def add_total_price(df):
    """Function 1: Calculate total price for each order"""
    df['total_price'] = df['unit_price'] * df['quantity']
    return df

def add_net_price(df):
    """Function 2: Calculate net price after discount"""
    df['net_price'] = df['total_price'] * (1 - df['discount'])
    return df

df=(df.pipe(add_total_price).pipe(add_net_price))
df.head()
```

### creating Index as order_date on dataframe

```python
# Convert order_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'])
df_indexed=df.set_index("order_date")
df_indexed.head()
```

## calculating monthly revenue

```python
# Group by month using resample
monthly_revenue = df_indexed['net_price'].resample('ME').sum()
print("\nMonthly Revenue:")
print(monthly_revenue)
```

## checking if monthly revenue is increasing month by month or not 

```python
monthly_revenue.index.is_monotonic_increasing
```
