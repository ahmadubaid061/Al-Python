```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('../healthcare_dataset.csv')
df.head()
```

## 1: Axis level Plot

```python
df=df.sample(500)
sns.scatterplot(data=df,x='treatment_cost',y='recovery_score',hue='visit_type') #hue means diffreent colors for each type
plt.title('Treatment Cost vs Recovery score')
plt.show()
```

## 2: Figure level plot

```python
sns.relplot(data=df,x='treatment_cost',y='recovery_score',hue='visit_type',col='gender') 
#col means it will draw two graphs for each gender
sns.set_theme(style='whitegrid')  #it will create rid lines in the white background
```
