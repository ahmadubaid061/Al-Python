## Corelation with Heatmap

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv('../healthcare_dataset.csv')
#we need only numeric columns for corelation so
df=df[['length_of_stay_days','treatment_cost','recovery_score','readmission_risk']]

df.head()
```

## corelation matrix

```python
corr_matrix=df.corr()
corr_matrix
```

## visualizing with heatmap

```python
sns.heatmap(corr_matrix,annot=True,cmap='coolwarm')
# annot means annotaion (displays values inside the sub boxes)
#cmap means color scheme
plt.title("Correlation visualization heatmap")
plt.xticks(rotation=30)
plt.show()
```

## masking upper triangle
- the matrix contains two identical triangles if we observe
- so we need only one of those to understand the data correlation 


```python
import numpy as np
mask=np.triu(np.ones_like(corr_matrix,dtype=bool))

#now plotting with heatmap
sns.set_theme(style='dark')
sns.heatmap(corr_matrix,mask=mask,annot=True,cmap='RdBu',)
plt.title("Correlation Heatmap masked")
plt.show()
```
