# Categorical Data Visualization
- two main categories
- 1: aggregate tendancy (Bar plot and count plot)
- 2: Distribution     (Box plot, voilin Plot,swarm plot,strip plot)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df=pd.read_csv('../healthcare_dataset.csv')
df.head(3)
```

## Aggregate count plot

```python
df=df.sample(500)
sns.countplot(data=df,x='treatment_type')
plt.title("Number of visits by visit type")
plt.show()
```

## Aggregate bar plot 

```python
sns.barplot(data=df,x='department',y='treatment_cost')
plt.xticks(rotation=30) #the x labels are rotated to 30 degree to reduce space consumtion
plt.show()
```

## Distribution based categorical plots

```python
# 1: Box plot
sns.boxplot(data=df,x='department',y='recovery_score')
plt.xticks(rotation=30)
plt.show()
```

```python
# 2: voilin plot
sns.violinplot(data=df,x='department',y='recovery_score')
plt.xticks(rotation=30)
plt.show()
```

```python
# 1: strip plot
sns.stripplot(data=df,x='department',y='recovery_score',jitter=True)
plt.xticks(rotation=30)
plt.show()
```

```python
#side by side box plot vs strip plot

sns.boxplot(data=df,x='department',y='recovery_score')
sns.stripplot(data=df,x='department',y='recovery_score',jitter=True,color='orange',alpha=0.4)
# -------------------------------this below code is just for designing the legend
import matplotlib.patches as mpatches
import matplotlib.lines as mlines
box_patch = mpatches.Patch(facecolor='lightblue', edgecolor='black', label='Box Plot distribution')
strip_dot = mlines.Line2D([], [], color='orange', marker='o', linestyle='None', alpha=0.4, label='Strip plot distribution')
plt.legend(handles=[box_patch, strip_dot])

# --------------------------------------------------
plt.xticks(rotation=30)
plt.show()
```
