```python
import pandas as pd
```

## Creating data set

```python
data = {
    'Name': ['Samiullah Khan', 'Gulalai Khattak', 'Sher Afzal', 'Zarlasht', 'Asfandyar Khan', 'Meena Gul'],
    'Age': [30, 26, 45, 33, 28, 39],
    'City': ['Peshawar', 'Mardan', 'Kabal', 'Islamabad', 'Lahore', 'Swat'],
    'Salary': [48000, 41000, 65000, 38000, 52000, 56000]
}

df=pd.DataFrame(data)
print(df)
```

## .loc function requires two set of arguments 
- (starting row, endin row)
- (starting column, ending column)  

```python
# columns Label shold be given for loc funciton
sub_set=df.loc[0:3,['Name','City']] #row 0 to row 3 and column Name to column City
print(sub_set)
```

```python
# iloc function works the same but it only accepts indeces not column labels
sub_set2=df.iloc[0:3,0:3] #row 0 to 2 column 0 to 2  here 2(index) means 3rd row
print(sub_set2)
```

```python
# using lock we can also assign new values to some columns
# suppose people with age > 30 there salary should be 65000 each
df.loc[df['Age'] > 30, 'Salary'] = 65000
print(df)  #now salary of those whose age is > 30 are 65000
```

## Boolean function

```python
high_salaries=df[df['Salary']>50000] #only those rows with salary > 50000
print(high_salaries)
```
